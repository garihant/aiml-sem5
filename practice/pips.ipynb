```python
from google import genai

client = genai.Client()
```

```python
x = "They lost my luggage and the flight was delayed for hours"
prevv = "Return one word for this statement, Positive, Negative or Neutral?"
response = client.models.generate_content(
    model="gemini-2.5-flash", contents=f"{x} {prevv}"
)
print(response.text)
```
