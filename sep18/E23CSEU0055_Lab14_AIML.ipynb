```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score
```

 Load the Breast Cancer dataset from scikit-learn. 

```python
data = load_breast_cancer()
x = data.data
y = data.target
```

Split the dataset into training and testing sets. 

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, random_state=12, test_size=0.2)
```

Standardize the dataset using StandardScaler.

```python
sr = StandardScaler()
x_std_train = sr.fit_transform(x_train)
x_std_test = sr.fit_transform(x_test)
```

Select different activation functions (ReLU, tanh, logistic) for 
experimentation. 
- Train the neural network with MLPClassifier for each combination of activation function 
and optimizer. 
- Record the accuracy for each run and compare the results. 

```python
supreme = ['relu', 'tanh', 'logistic']
results = []

for landman in supreme:
    clf = MLPClassifier(hidden_layer_sizes=(64, 32, 16, 8), activation=landman, max_iter=750, random_state=22)
    clf.fit(x_std_train, y_train)
    y_pred = clf.predict(x_std_test)
    acc = accuracy_score(y_test, y_pred)
    results.append({'activation': landman, 'accuracy': acc})

results_df = pd.DataFrame(results)
print(results_df.sort_values(by='accuracy', ascending=False))
```
